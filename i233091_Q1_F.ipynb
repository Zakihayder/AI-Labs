```python
graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['E'],
    'D': [],
    'E': []
}

class SearchProblem:
    def __init__(self, graph, initial, goal):
        self.graph   = graph
        self.initial = initial
        self.goal    = goal

    def actions(self, state):
        return self.graph.get(state, [])

    def result(self, state, action):
        return action   # action IS the next state

    def goal_test(self, state):
        return state == self.goal

    def path_cost(self, cost, state1, action, state2):
        return cost + 1   # unit cost

# T0501: Formal state-space model using adjacency list
# Concept: Adjacency list, state-space representation, problem formulation components
problem = SearchProblem(graph, 'A', 'E')
print("Initial State:", problem.initial)
print("Goal State:", problem.goal)
print("Actions from A:", problem.actions('A'))
print("Result of A + move to C:", problem.result('A', 'C'))
print("Goal test on E:", problem.goal_test('E'))
print("Graph adjacency list:", graph)
```

```python
from collections import deque

tree = {'S': ['A','B'], 'A': ['C'], 'B': [], 'C': []}
graph = {'S': ['A','B'], 'A': ['C'], 'B': ['C'], 'C': ['B']}

def bfs_tree(adj, start, goal):
    """BFS on a tree — no visited set needed (no cycles)."""
    queue = deque([[start]])
    while queue:
        path = queue.popleft()
        node = path[-1]
        print(f"Visiting: {node}")
        if node == goal:
            print(f"Goal found! Path: {' -> '.join(path)}\n")
            return
        for nbr in adj.get(node, []):
            queue.append(path + [nbr])

def bfs_graph_no_visited(adj, start, goal, max_steps=12):
    """BFS on graph WITHOUT visited set — demonstrates infinite loop."""
    queue = deque([[start]])
    steps = 0
    while queue and steps < max_steps:
        path = queue.popleft()
        node = path[-1]
        steps += 1
        print(f"Step {steps}: Visiting {node}")
        if node == goal:
            print(f"Goal found! Path: {' -> '.join(path)}\n")
            return
        for nbr in adj.get(node, []):
            queue.append(path + [nbr])
    print("Stopped — infinite loop!\n")

def bfs_graph_visited(adj, start, goal):
    """BFS on graph WITH visited set — correct."""
    visited = set()
    queue   = deque([[start]])
    while queue:
        path = queue.popleft()
        node = path[-1]
        if node in visited:
            print(f"  (Skipping {node} — already visited)")
            continue
        visited.add(node)
        print(f"Visiting: {node}")
        if node == goal:
            print(f"Goal found! Path: {' -> '.join(path)}\n")
            return
        for nbr in adj.get(node, []):
            queue.append(path + [nbr])

print("=== Part 1: BFS on TREE (no visited set) ===")
bfs_tree(tree, 'S', 'C')

# T0502: BFS on tree (no visited set), graph (no visited set), graph (with visited set)
# Concept: BFS, visited set, infinite loops in cyclic graphs, tree vs graph search
print("=== Part 2: BFS on GRAPH (no visited set) — infinite loop ===")
bfs_graph_no_visited(graph, 'S', 'C')

print("=== Part 3: BFS on GRAPH (WITH visited set) ===")
bfs_graph_visited(graph, 'S', 'C')
```

```python
graph_dfs = {
    'S': ['A'],
    'A': ['B'],
    'B': ['C', 'G'],
    'C': [],
    'G': []
}

def dfs(graph, start, goal):
    stack   = [(start, [start])]
    visited = set()
    print("DFS exploration:")
    while stack:
        node, path = stack.pop()
        if node in visited:
            continue
        visited.add(node)
        print(f"  Expanding: {node}")
        if node == goal:
            print(f"Goal found! Path: {' -> '.join(path)}")
            return path
        for nbr in reversed(graph.get(node, [])):  # reversed to match natural order
            if nbr not in visited:
                stack.append((nbr, path + [nbr]))
    print("Goal not reachable.")
    return None
    
# T0503: DFS implementation using explicit LIFO stack with visited set
# Concept: DFS, LIFO stack, visited set, deep path exploration
path = dfs(graph_dfs, 'S', 'G')
```

```python
import heapq

graph_ucs = {
    'S': [('A', 1), ('B', 5)],
    'A': [('G', 2)],
    'B': [('G', 1)],
    'G': []
}

def ucs(graph, start, goal):
    heap    = [(0, start, [start])]   # (cost, node, path)
    visited = set()
    print("UCS exploration (cost, node):")
    while heap:
        cost, node, path = heapq.heappop(heap)
        if node in visited:
            continue
        visited.add(node)
        print(f"  Expanding: {node} (cost so far: {cost})")
        if node == goal:
            print(f"\nOptimal path: {' -> '.join(path)}")
            print(f"Total cost: {cost}")
            return path, cost
        for nbr, edge_cost in graph.get(node, []):
            if nbr not in visited:
                heapq.heappush(heap, (cost + edge_cost, nbr, path + [nbr]))
    print("Goal not reachable.")
    return None, float('inf')

# T0504: Uniform Cost Search using heapq priority queue
# Concept: Priority queue (min-heap), cumulative cost, optimal path
ucs(graph_ucs, 'S', 'G')
```

```python
from collections import deque

graph_bfs = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['G'],
    'C': [], 'D': [], 'G': []
}

def bfs(graph, start, goal):
    queue    = deque([start])
    visited  = {start}
    parent   = {start: None}
    print("BFS expansion order:")
    while queue:
        node = queue.popleft()
        print(f"  Visiting: {node}")
        if node == goal:
            # Reconstruct path
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            path.reverse()
            print(f"Goal found! Path: {' -> '.join(path)}")
            return path
        for nbr in graph.get(node, []):
            if nbr not in visited:
                visited.add(nbr)
                parent[nbr] = node
                queue.append(nbr)
    print("Goal not reachable.")
    return None

# T0505: BFS with parent pointers for path reconstruction
# Concept: FIFO queue, explored set, parent pointers, path reconstruction
bfs(graph_bfs, 'S', 'G')
```

```python
import heapq
from collections import deque

campus = {
    'Hostel':    [('Cafeteria', 2)],
    'Cafeteria': [('Library', 3), ('Lab', 1)],
    'Library':   [],
    'Lab':       []
}

start, goal = 'Hostel', 'Lab'
adj_unweighted = {n: [nb for nb, _ in nbrs] for n, nbrs in campus.items()}

# BFS
def bfs_nav(graph, s, g):
    q = deque([[s]]); vis = {s}
    while q:
        path = q.popleft(); node = path[-1]
        if node == g: return path
        for nb in graph.get(node, []):
            if nb not in vis: vis.add(nb); q.append(path+[nb])
    return None

# DFS
def dfs_nav(graph, s, g):
    stack = [(s, [s])]; vis = set()
    while stack:
        node, path = stack.pop()
        if node in vis: continue
        vis.add(node)
        if node == g: return path
        for nb in reversed(graph.get(node, [])):
            if nb not in vis: stack.append((nb, path+[nb]))
    return None

# UCS
def ucs_nav(graph, s, g):
    heap = [(0, s, [s])]; vis = set()
    while heap:
        cost, node, path = heapq.heappop(heap)
        if node in vis: continue
        vis.add(node)
        if node == g: return path, cost
        for nb, w in graph.get(node, []):
            if nb not in vis: heapq.heappush(heap, (cost+w, nb, path+[nb]))
    return None, 0

bfs_path = bfs_nav(adj_unweighted, start, goal)
dfs_path = dfs_nav(adj_unweighted, start, goal)
ucs_path, ucs_cost = ucs_nav(campus, start, goal)

# T0506: Campus navigation using BFS, DFS, and UCS
# Concept: Real-world search application, comparison of BFS/DFS/UCS on same graph
print(f"BFS path ({start} -> {goal}): {' -> '.join(bfs_path)}")
print(f"DFS path ({start} -> {goal}): {' -> '.join(dfs_path)}")
print(f"UCS path ({start} -> {goal}): {' -> '.join(ucs_path)}, cost={ucs_cost}")
```

```python
import heapq

def h(node):
    return heuristic4[node]

# Compute actual shortest path (Dijkstra) from each node to G
def dijkstra(graph, start, goal):
    heap = [(0, start)]
    dist = {start: 0}
    while heap:
        cost, node = heapq.heappop(heap)
        if node == goal: return cost
        for nbr, w in graph.get(node, []):
            nc = cost + w
            if nc < dist.get(nbr, float('inf')):
                dist[nbr] = nc
                heapq.heappush(heap, (nc, nbr))
    return float('inf')

nodes = list(graph4.keys())
print(f"{'Node':<8} {'h(n)':<10} {'Actual cost to G':<20} {'Admissible?'}")
print("-"*50)
for node in nodes:
    actual = dijkstra(graph4, node, 'G')
    admissible = h(node) <= actual
    print(f"  {node:<6} {h(node):<10} {actual:<20} {'YES' if admissible else 'NO'}")

# Consistency check: h(n) <= cost(n,m) + h(m) for all edges
print("\nConsistency check (h(n) <= cost(n,m) + h(m)):")
consistent = True
for node, neighbors in graph4.items():
    for nbr, cost in neighbors:
        if h(node) > cost + h(nbr):
            print(f"  INCONSISTENT: {node}->{nbr}: h({node})={h(node)} > {cost}+h({nbr})={cost+h(nbr)}")
            consistent = False
            
# T0601: Verify admissibility and consistency of heuristic; compare with actual shortest paths
# Concept: Admissibility, consistency (triangle inequality), heuristic vs actual cost
# A heuristic is admissible if it never overestimates; consistent if h(n) ≤ cost(n,m) + h(m).
if consistent:
    print("  All edges satisfy consistency. Heuristic is CONSISTENT.")
```

```python
import heapq

def gbfs(graph, heuristic, start, goal):
    heap = [(heuristic[start], start, [start])]
    visited = set()
    print("GBFS expansion order:")
    while heap:
        h_val, node, path = heapq.heappop(heap)
        if node in visited: continue
        visited.add(node)
        print(f"  Expanding: {node} (h={h_val})")
        if node == goal:
            cost = sum(
                next(w for nb, w in graph[path[i]] if nb == path[i+1])
                for i in range(len(path)-1)
            )
            print(f"\nPath found: {' -> '.join(path)}")
            print(f"Total path cost: {cost}")
            return path, cost
        for nbr, _ in graph.get(node, []):
            if nbr not in visited:
                heapq.heappush(heap, (heuristic[nbr], nbr, path+[nbr]))
    return None, float('inf')

# T0602: Greedy Best-First Search using f(n)=h(n)
# Concept: GBFS, f(n)=h(n), priority queue on heuristic, greedy node selection
gbfs_path, gbfs_cost = gbfs(graph4, heuristic4, 'A', 'G')
```

```python
import random

random.seed(42)
POP_SIZE, CHROM_LEN, GENERATIONS, MUTATION_RATE = 6, 10, 5, 0.1

def fitness(chrom): return sum(chrom)

def roulette_select(pop, fitnesses):
    total = sum(fitnesses)
    r = random.uniform(0, total)
    cumulative = 0
    for chrom, f in zip(pop, fitnesses):
        cumulative += f
        if cumulative >= r: return chrom
    return pop[-1]

def single_point_crossover(p1, p2):
    pt = random.randint(1, CHROM_LEN-1)
    return p1[:pt]+p2[pt:], p2[:pt]+p1[pt:]

def mutate(chrom, rate=MUTATION_RATE):
    return [bit^1 if random.random()<rate else bit for bit in chrom]

population = [[random.randint(0,1) for _ in range(CHROM_LEN)] for _ in range(POP_SIZE)]

for gen in range(GENERATIONS):
    fitnesses = [fitness(c) for c in population]
    best = population[fitnesses.index(max(fitnesses))]
    print(f"Gen {gen+1}: Best fitness={max(fitnesses)}, Chromosome={best}")
    new_pop = []
    while len(new_pop) < POP_SIZE:
        p1 = roulette_select(population, fitnesses)
        p2 = roulette_select(population, fitnesses)
        c1, c2 = single_point_crossover(p1, p2)
        new_pop.extend([mutate(c1), mutate(c2)])
    population = new_pop[:POP_SIZE]

# T0801: Basic GA to maximize 1s in a binary string of length 10
# Concept: GA fundamentals — chromosome, population, fitness, roulette selection, crossover, mutation
# Maximizes number of 1s in a 10-bit binary string over 5 generations.
fitnesses = [fitness(c) for c in population]
print(f"\nFinal best: {population[fitnesses.index(max(fitnesses))]} (fitness={max(fitnesses)})")
```

```python
import random, math
import matplotlib.pyplot as plt

random.seed(0)
POP_SIZE, GENERATIONS = 6, 10
f = lambda x: x * math.sin(10 * math.pi * x) + 1.0

def tournament_select(pop, fitnesses, k=2):
    idx = random.sample(range(len(pop)), k)
    return pop[max(idx, key=lambda i: fitnesses[i])]

def arithmetic_crossover(p1, p2):
    a = random.random()
    return a*p1 + (1-a)*p2

def gaussian_mutate(x):
    return max(0.0, min(1.0, x + random.gauss(0, 0.05)))

population = [random.random() for _ in range(POP_SIZE)]
best_fitness_history = []

for gen in range(GENERATIONS):
    fitnesses = [f(x) for x in population]
    best_x = population[fitnesses.index(max(fitnesses))]
    best_f = max(fitnesses)
    best_fitness_history.append(best_f)
    print(f"Gen {gen+1}: Best x={best_x:.4f}, f(x)={best_f:.4f}")
    new_pop = []
    while len(new_pop) < POP_SIZE:
        p1 = tournament_select(population, fitnesses)
        p2 = tournament_select(population, fitnesses)
        child = arithmetic_crossover(p1, p2)
        new_pop.append(gaussian_mutate(child))
    population = new_pop

# T0802: GA for maximizing f(x) = x*sin(10*pi*x) + 1 over x in [0,1]
# Concept: Real-valued chromosome, arithmetic crossover, Gaussian mutation, tournament selection
plt.figure(figsize=(8,4))
plt.plot(range(1, GENERATIONS+1), best_fitness_history, 'b-o')
plt.title("Task 0802 — Best Fitness per Generation"); plt.xlabel("Generation"); plt.ylabel("Best f(x)")
plt.grid(True); plt.tight_layout(); plt.show()
```

```python
import random
random.seed(7)

items = [(2,3),(3,4),(4,5),(5,8)]   # (weight, value)
MAX_WEIGHT = 8
POP_SIZE, GENERATIONS = 6, 10

def fitness_knapsack(chrom):
    w = sum(items[i][0] for i in range(4) if chrom[i])
    v = sum(items[i][1] for i in range(4) if chrom[i])
    return v if w <= MAX_WEIGHT else 0

def roulette(pop, fits):
    total = sum(fits) or 1
    r = random.uniform(0, total); cum = 0
    for c, f in zip(pop, fits):
        cum += f
        if cum >= r: return c
    return pop[-1]

def crossover(p1, p2):
    pt = random.randint(1,3)
    return p1[:pt]+p2[pt:], p2[:pt]+p1[pt:]

def mutate(c):
    return [bit^1 if random.random()<0.1 else bit for bit in c]

pop = [[random.randint(0,1) for _ in range(4)] for _ in range(POP_SIZE)]

for gen in range(GENERATIONS):
    fits = [fitness_knapsack(c) for c in pop]
    best = pop[fits.index(max(fits))]
    w = sum(items[i][0] for i in range(4) if best[i])
    v = sum(items[i][1] for i in range(4) if best[i])
    print(f"Gen {gen+1}: Best={best}, Weight={w}, Value={v}")
    new_pop = []

    # T0803: GA for 0/1 Knapsack — maximize value with weight <= 8
    # Concept: Binary encoding, constraint-based fitness (weight penalty), combinatorial optimization
    # Evolves item selection to maximize value while staying within weight limit.
    while len(new_pop) < POP_SIZE:
        p1 = roulette(pop, fits); p2 = roulette(pop, fits)
        c1, c2 = crossover(p1, p2)
        new_pop.extend([mutate(c1), mutate(c2)])
    pop = new_pop[:POP_SIZE]
```

```python
import random
random.seed(3)

cities = ['A','B','C','D','E']
dist = {
    ('A','B'):2,('A','C'):9,('A','D'):10,('A','E'):7,
    ('B','C'):6,('B','D'):4,('B','E'):3,
    ('C','D'):8,('C','E'):5,
    ('D','E'):6
}

def route_dist(route):
    d = 0
    for i in range(len(route)):
        a, b = route[i], route[(i+1)%len(route)]
        d += dist.get((a,b), dist.get((b,a), 0))
    return d

def fitness_tsp(route): return 1 / route_dist(route)

def order_crossover(p1, p2):
    n = len(p1); a, b = sorted(random.sample(range(n), 2))
    child = [None]*n; child[a:b] = p1[a:b]
    fill = [x for x in p2 if x not in child]
    j = 0
    for i in range(n):
        if child[i] is None:
            child[i] = fill[j]; j += 1
    return child

def swap_mutate(route):
    r = route[:]
    if random.random() < 0.2:
        i, j = random.sample(range(len(r)), 2)
        r[i], r[j] = r[j], r[i]
    return r

def tournament(pop, fits):
    i, j = random.sample(range(len(pop)), 2)
    return pop[i] if fits[i] > fits[j] else pop[j]

pop = [random.sample(cities, len(cities)) for _ in range(6)]

for gen in range(10):
    fits = [fitness_tsp(r) for r in pop]
    best = pop[fits.index(max(fits))]
    print(f"Gen {gen+1}: Best route={best}, Distance={route_dist(best)}")
    new_pop = []

    # T0804: GA for TSP with 5 cities using Order Crossover (OX)
    # Concept: Permutation encoding, order crossover (OX), swap mutation, TSP
    # Evolves city visit orderings to minimize total tour distance across 5 cities.
    while len(new_pop) < 6:
        p1 = tournament(pop, fits); p2 = tournament(pop, fits)
        child = swap_mutate(order_crossover(p1, p2))
        new_pop.append(child)
    pop = new_pop
```

```python
import random
random.seed(5)

def target_F(a, b, c): return int((a and b) or (b and c))

# Fitness: number of truth table rows where chromosome gives correct F output
truth_table = [(a,b,c) for a in [0,1] for b in [0,1] for c in [0,1]]

def fitness_logic(chrom):
    a, b, c = chrom
    correct = sum(1 for ta,tb,tc in truth_table if target_F(ta,tb,tc)==target_F(a,b,c))
    return correct

pop = [[random.randint(0,1) for _ in range(3)] for _ in range(6)]

for gen in range(6):
    fits = [fitness_logic(c) for c in pop]
    best = pop[fits.index(max(fits))]
    print(f"Gen {gen+1}: Fitnesses={fits}, Best={best} (score={max(fits)})")
    new_pop = []
    total = sum(fits) or 1
    while len(new_pop) < 6:
        # Roulette selection
        def roulette_pick():
            r = random.uniform(0, total); cum = 0
            for c, f in zip(pop, fits):
                cum += f
                if cum >= r: return c
            return pop[-1]
        p1, p2 = roulette_pick(), roulette_pick()

        # T0805: GA for symbolic/logic optimization — maximize F(a,b,c) = (a AND b) OR (b AND c)
        # Concept: Boolean chromosome, truth-table fitness, symbolic optimization
        # Evolves [a,b,c] bit strings to maximize correct evaluations of F(a,b,c) = (a AND b) OR (b AND c).
        pt = random.randint(1, 2)
        child = [bit^1 if random.random()<0.15 else bit for bit in (p1[:pt]+p2[pt:])]
        new_pop.append(child)
    pop = new_pop
```

```python
import random
random.seed(9)

TARGET = 50
POP_SIZE, GENERATIONS = 6, 10

def fitness_eq(chrom):
    x, y = chrom
    return 1 / (1 + abs(x**2 + y**2 - TARGET))

def tournament(pop, fits):
    i, j = random.sample(range(len(pop)), 2)
    return pop[i] if fits[i] > fits[j] else pop[j]

def avg_crossover(p1, p2):
    return [round((p1[0]+p2[0])/2), round((p1[1]+p2[1])/2)]

def mutate_eq(chrom):
    c = chrom[:]
    for i in range(len(c)):
        if random.random() < 0.2:
            c[i] = max(0, min(10, c[i] + random.choice([-1,1])))
    return c

pop = [[random.randint(0,10), random.randint(0,10)] for _ in range(POP_SIZE)]

for gen in range(GENERATIONS):
    fits = [fitness_eq(c) for c in pop]
    best = pop[fits.index(max(fits))]
    print(f"Gen {gen+1}: Best [x,y]={best}, x²+y²={best[0]**2+best[1]**2}, fitness={max(fits):.4f}")
    new_pop = []
    while len(new_pop) < POP_SIZE:
        p1 = tournament(pop, fits); p2 = tournament(pop, fits)
        child = mutate_eq(avg_crossover(p1, p2))
        new_pop.append(child)
    pop = new_pop
# T0806: GA to find integers x,y in [0,10] satisfying x^2 + y^2 = 50
#Concept: Integer chromosome, proximity-based fitness, average crossover, ±1 mutation
#Evolves integer pairs (x,y) to satisfy the equation x² + y² = 50.
fits = [fitness_eq(c) for c in pop]
best = pop[fits.index(max(fits))]
print(f"\nFinal solution: x={best[0]}, y={best[1]}, x²+y²={best[0]**2+best[1]**2} (target=50)")
```
